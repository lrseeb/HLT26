### Making the data to list of lists:

```python
with open("enwiki-20220301-sample-tokenized.txt") as data:
    sentences: list["list"] = []
    for line in data.readlines():
        sentences.append(line.split())
```

### Training the model:

```python
from gensim.models import Word2Vec

# train model

model = Word2Vec(
    sentences=sentences,
    sg=0,
    vector_size=40,
    window=2,
    min_count=1,
    workers=1,
    epochs=100
)
```

```python
context = ["language", "modeling", "objective"]
print("Context:", context)
pred = model.predict_output_word(context, topn=5)
print("Top predictions:")
for w,s in pred:
    print(f"{w}: {s:.5f}")
```

```python
print("good:", model.wv.most_similar('good', topn=5))
print("bad:", model.wv.most_similar('bad', topn=5))
print("dog:", model.wv.most_similar('dog', topn=5))
print("cat:", model.wv.most_similar('cat', topn=5))
print("king:", model.wv.most_similar('king', topn=5))
print("queen:", model.wv.most_similar('queen', topn=5))
```
