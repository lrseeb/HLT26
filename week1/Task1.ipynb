Exercise task 1 (Basic): Subword tokenization

```python
# Importing dependencies
import wordfreq as wf
from transformers import AutoTokenizer
from collections import Counter
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
```

```python
wordlist: list[str] = wf.top_n_list(lang="en", n=100000)
```

```python
wordlist_as_string: str = ' '.join(wordlist)
```

```python
tokenized_list: list["str"] = tokenizer.tokenize(wordlist_as_string)
```

```python
print(len(tokenized_list))
```

```python
for i in range(20):
    print(tokenized_list[i])
```

```python
token_counter = Counter(tokenized_list)

print("Number of tokens in total:", token_counter.total())
print("Most common tokens:")
for item in token_counter.most_common(20):
  print(item)
print("Vocabulary size:", len(token_counter))
```
